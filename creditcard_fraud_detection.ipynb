```python

# ============================================================
# PROJECT: CREDIT CARD FRAUD DETECTION USING MACHINE LEARNING
# ALGORITHM: RANDOM FOREST CLASSIFIER
# PLATFORM: GOOGLE COLAB
# ============================================================


# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import zipfile

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")


# ============================================================
# 2. UPLOAD DATASET
# ============================================================

print("Please upload creditcard.csv")

uploaded = files.upload()

df = pd.read_csv('creditcard.csv')

print("\nDataset loaded successfully!")


# ============================================================
# 3. DATASET EXPLORATION
# ============================================================

print("\n========== DATASET INFORMATION ==========")

print("\nFirst Five Rows:")
display(df.head())

print("\nLast Five Rows:")
display(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Description:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nTransaction Distribution:")
print(df['Class'].value_counts())


# ============================================================
# 4. DATA ANALYSIS
# ============================================================

total_transactions = len(df)

genuine_transactions = (df['Class'] == 0).sum()

fraud_transactions = (df['Class'] == 1).sum()

fraud_percentage = (fraud_transactions / total_transactions) * 100

print("\n========== TRANSACTION ANALYSIS ==========")

print("Total Transactions:", total_transactions)

print("Genuine Transactions:", genuine_transactions)

print("Fraudulent Transactions:", fraud_transactions)

print("Fraud Percentage:", round(fraud_percentage, 3), "%")


# ============================================================
# 5. VISUALIZATION 1: FRAUD VS GENUINE TRANSACTIONS
# ============================================================

plt.figure(figsize=(8,5))

sns.countplot(x='Class', data=df)

plt.title("Fraud vs Genuine Transactions")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")

plt.xticks([0,1], ['Genuine', 'Fraud'])

plt.savefig("transaction_distribution.png")

plt.show()


# ============================================================
# 6. VISUALIZATION 2: TRANSACTION PERCENTAGE
# ============================================================

plt.figure(figsize=(7,7))

plt.pie(
    [genuine_transactions, fraud_transactions],
    labels=['Genuine', 'Fraud'],
    autopct='%1.3f%%',
    startangle=90
)

plt.title("Percentage of Fraud and Genuine Transactions")

plt.savefig("transaction_percentage.png")

plt.show()


# ============================================================
# 7. VISUALIZATION 3: TRANSACTION AMOUNT DISTRIBUTION
# ============================================================

plt.figure(figsize=(10,5))

sns.histplot(
    data=df,
    x='Amount',
    bins=50,
    kde=True
)

plt.title("Transaction Amount Distribution")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.savefig("amount_distribution.png")

plt.show()


# ============================================================
# 8. VISUALIZATION 4: FRAUD VS GENUINE AMOUNT
# ============================================================

plt.figure(figsize=(9,5))

sns.boxplot(
    x='Class',
    y='Amount',
    data=df
)

plt.title("Transaction Amount by Class")
plt.xlabel("0 = Genuine, 1 = Fraud")
plt.ylabel("Transaction Amount")

plt.savefig("amount_comparison.png")

plt.show()


# ============================================================
# 9. VISUALIZATION 5: CORRELATION HEATMAP
# ============================================================

correlation = df.corr()

plt.figure(figsize=(16,12))

sns.heatmap(
    correlation,
    cmap='coolwarm',
    linewidths=0.1
)

plt.title("Correlation Heatmap")

plt.savefig("correlation_heatmap.png")

plt.show()


# ============================================================
# 10. DATA PREPROCESSING
# ============================================================

print("\n========== DATA PREPROCESSING ==========")

# Check missing values

print("Missing Values:", df.isnull().sum().sum())

# Remove duplicate records

df = df.drop_duplicates().copy()

print("Dataset shape after removing duplicates:", df.shape)

# Separate independent and dependent variables

X = df.drop('Class', axis=1)

y = df['Class']

print("Input Features Shape:", X.shape)

print("Target Variable Shape:", y.shape)


# ============================================================
# 11. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Data:", X_train.shape)

print("Testing Data:", X_test.shape)


# ============================================================
# 12. RANDOM FOREST CLASSIFIER
# ============================================================

print("\n========== MODEL TRAINING ==========")

model = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced_subsample',
    random_state=42,
    n_jobs=-1
)

# Train the model

model.fit(X_train, y_train)

print("Random Forest Model Trained Successfully!")


# ============================================================
# 13. MODEL PREDICTION
# ============================================================

y_pred = model.predict(X_test)

y_probability = model.predict_proba(X_test)[:, 1]

print("\nPredictions Completed Successfully!")


# ============================================================
# 14. MODEL PERFORMANCE EVALUATION
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(y_test, y_pred, zero_division=0)

recall = recall_score(y_test, y_pred, zero_division=0)

f1 = f1_score(y_test, y_pred, zero_division=0)

roc_auc = roc_auc_score(y_test, y_probability)

pr_auc = average_precision_score(y_test, y_probability)

print("\n========== MODEL PERFORMANCE ==========")

print("Accuracy :", round(accuracy * 100, 2), "%")

print("Precision:", round(precision * 100, 2), "%")

print("Recall   :", round(recall * 100, 2), "%")

print("F1 Score :", round(f1 * 100, 2), "%")

print("ROC-AUC  :", round(roc_auc, 4))

print("PR-AUC   :", round(pr_auc, 4))

print("\nClassification Report:")

print(classification_report(y_test, y_pred))


# ============================================================
# 15. VISUALIZATION 6: CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7,5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted Values")
plt.ylabel("Actual Values")

plt.savefig("confusion_matrix.png")

plt.show()


# ============================================================
# 16. VISUALIZATION 7: FEATURE IMPORTANCE
# ============================================================

importance = pd.Series(
    model.feature_importances_,
    index=X.columns
)

importance = importance.sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(10,6))

importance.sort_values().plot(kind='barh')

plt.title("Top 10 Important Features")

plt.xlabel("Importance Score")
plt.ylabel("Features")

plt.savefig("feature_importance.png")

plt.show()


# ============================================================
# 17. VISUALIZATION 8: PERFORMANCE COMPARISON
# ============================================================

metrics = {
    'Accuracy': accuracy,
    'Precision': precision,
    'Recall': recall,
    'F1 Score': f1
}

plt.figure(figsize=(9,5))

plt.bar(
    metrics.keys(),
    [value * 100 for value in metrics.values()]
)

plt.title("Random Forest Model Performance")

plt.xlabel("Evaluation Metrics")
plt.ylabel("Performance Score (%)")

plt.ylim(0,100)

plt.savefig("performance_comparison.png")

plt.show()


# ============================================================
# 18. SAVE PERFORMANCE RESULTS
# ============================================================

results = pd.DataFrame({
    'Metric': [
        'Accuracy',
        'Precision',
        'Recall',
        'F1 Score',
        'ROC-AUC',
        'PR-AUC'
    ],

    'Score': [
        accuracy,
        precision,
        recall,
        f1,
        roc_auc,
        pr_auc
    ]
})

print("\n========== FINAL RESULTS ==========")

display(results)

results.to_csv("model_performance.csv", index=False)


# ============================================================
# 19. FINAL CONCLUSION
# ============================================================

print("\n==============================================")
print("CREDIT CARD FRAUD DETECTION PROJECT")
print("==============================================")

print("\nAlgorithm Used: Random Forest Classifier")

print("Dataset: Credit Card Transactions")

print("Total Transactions:", len(df))

print("Genuine Transactions:", (df['Class'] == 0).sum())

print("Fraudulent Transactions:", (df['Class'] == 1).sum())

print("\nModel Evaluation:")

print("Accuracy:", round(accuracy * 100, 2), "%")

print("Precision:", round(precision * 100, 2), "%")

print("Recall:", round(recall * 100, 2), "%")

print("F1 Score:", round(f1 * 100, 2), "%")

print("ROC-AUC:", round(roc_auc, 4))

print("PR-AUC:", round(pr_auc, 4))

print("\nCONCLUSION:")

print("The Random Forest algorithm was used to identify")
print("fraudulent and genuine credit card transactions.")

print("The model was evaluated using different performance")
print("metrics and visualizations.")

print("This project demonstrates the application of")
print("Machine Learning in financial fraud detection.")

print("\nPROJECT COMPLETED SUCCESSFULLY!")


# ============================================================
# 20. SAVE ALL PROJECT OUTPUTS
# ============================================================

output_files = [
    "transaction_distribution.png",
    "transaction_percentage.png",
    "amount_distribution.png",
    "amount_comparison.png",
    "correlation_heatmap.png",
    "confusion_matrix.png",
    "feature_importance.png",
    "performance_comparison.png",
    "model_performance.csv"
]

with zipfile.ZipFile(
    "Credit_Card_Fraud_Results.zip",
    "w"
) as zip_file:

    for file in output_files:
        zip_file.write(file)

print("\nAll graphs and performance results saved!")

print("Downloading project results...")

files.download("Credit_Card_Fraud_Results.zip")

# ============================================================
# END OF PROJECT
# ============================================================
```
